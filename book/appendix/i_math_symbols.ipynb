# I. Mathematical Symbols and Notation

I collect the main conventions here so you can look up a symbol without losing your place in a chapter. Symbols get their meaning from the local definition: the same letter can stand for different quantities in different models.

## I.1 Signals, time, and frequency

| Symbol | Read it as | Meaning and units |
|---|---|---|
| $x(t)$ | x at time t | Continuous-time signal; e.g. mV |
| $x[n]$ | x at sample n | A discrete-time sequence |
| $t$ | Time | Seconds unless stated otherwise |
| $n$ | Sample index | Integer count |
| $f_s$ | Sampling frequency | Samples per second, in Hz |
| $T_s$ | Sampling interval | $1/f_s$, in seconds |
| $N$ | Number of samples | A count, unless locally redefined |
| $f$ | Frequency | Cycles per second, in Hz |
| $\omega$ | Angular frequency | $2\pi f$, in rad/s |
| $\Omega$ | Discrete angular frequency | $2\pi f/f_s$, in rad/sample |
| $\varphi$ | Phase | Usually radians |

For example, $x[n]=x(nT_s)$ evaluates a continuous waveform at the time of sample $n$. At $f_s=360$ Hz, sample 180 represents time 0.5 s.

## I.2 Comparisons and sets

| Symbol | Meaning | Example |
|---|---|---|
| $=$ | Equal | $T_s=1/f_s$ |
| $\approx$ | Approximately equal | $1/360\approx0.00278$ |
| $\ne$ | Not equal | $f_s\ne0$ |
| $\le$, $\ge$ | Less/greater than or equal | $0\le n<N$ |
| $\in$ | Is a member of | $n\in\mathbb{Z}$ |
| $\mathbb{R}$ | Real numbers | Real amplitudes |
| $\mathbb{C}$ | Complex numbers | Fourier coefficients |
| $\mathbb{Z}$ | Integers | Sample indices |
| $\lvert x\rvert$ | Absolute value or magnitude | $\lvert-2\rvert=2$ |
| $\propto$ | Proportional to | $y=cx$ for a fixed factor $c$ |

## I.3 Sums and calculus

$$
\sum_{n=0}^{N-1}x[n]=x[0]+x[1]+\cdots+x[N-1].
$$

The sum has $N$ terms. If $x=[1,2,3]$, it is 6; dividing by three gives a sample mean of 2.

| Symbol | Meaning |
|---|---|
| $\sum$ | Add the indexed terms |
| $\prod$ | Multiply the indexed terms |
| $dx/dt$ | Rate of change of $x$ with time |
| $\int_a^b x(t)\,dt$ | Accumulation between $a$ and $b$ |
| $\lim$ | Value approached in a stated limit |
| $\infty$ | Unbounded extent, not an ordinary finite number |

If $x$ is measured in mV and $t$ in seconds, its derivative is in mV/s and its time integral is in mV·s. Units provide a useful check on a derivation.

## I.4 Complex numbers and linear algebra

| Symbol | Meaning |
|---|---|
| $j$ | Imaginary unit, $j^2=-1$; Python writes `1j` |
| $z^*$ | Complex conjugate |
| $\operatorname{Re}\{z\}$, $\operatorname{Im}\{z\}$ | Real and imaginary parts |
| $e^{j\theta}$ | $\cos\theta+j\sin\theta$ |
| $\mathbf{x}$, $\mathbf{A}$ | Vector and matrix |
| $\mathbf{A}^{\mathsf T}$ | Transpose |
| $\mathbf{A}^{\mathsf H}$ | Conjugate transpose |
| $\lVert\mathbf{x}\rVert_2$ | Square root of the sum of squared magnitudes |

For $z=3+4j$, the conjugate is $3-4j$ and the magnitude is $\sqrt{3^2+4^2}=5$.

## I.5 Probability and signal operations

| Symbol | Meaning |
|---|---|
| $\mathbb{E}[X]$ | Expected value, a probability-weighted average |
| $\mu$ | Mean when defined that way in the model |
| $\sigma^2$ | Variance |
| $\hat{x}$ | Estimate of $x$ |
| $\bar{x}$ | Average of $x$ |
| $x*h$ | Convolution |
| $\delta[n]$ | Discrete unit impulse |
| $\delta(t)$ | Continuous Dirac impulse, a distribution |
| $X(f)$ | Fourier transform under the stated convention |
| $H$ | Frequency response or transfer function, depending on its argument |

The star between signals denotes convolution; a superscript star denotes conjugation. Also distinguish variance $\sigma^2$ from conductivity $\sigma$ in a volume-conductor model. Check definitions and units each time.
