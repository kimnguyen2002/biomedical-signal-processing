# H. Writing Equations with LaTeX

LaTeX commands describe how to typeset mathematics. A notebook renders those commands as equations. I use inline mathematics for a short symbol within a sentence and display mathematics when an expression needs its own line.

## H.1 Inline and display mathematics

Write `$f_s=360\,\mathrm{Hz}$` to display $f_s=360\,\mathrm{Hz}$. For a separate equation, use double-dollar delimiters:

```text
$$
T_s = \frac{1}{f_s}.
$$
```

$$
T_s = \frac{1}{f_s}.
$$

The sampling interval is the reciprocal of the sampling rate. At 360 Hz it is approximately 2.78 ms. `\frac` typesets a fraction; it does not calculate the value.

## H.2 Frequently used commands

| Purpose | Source | Result |
|---|---|---|
| Subscript | `x_n` | $x_n$ |
| Grouped subscript | `f_{max}` | $f_{max}$ |
| Superscript | `x^2` | $x^2$ |
| Fraction | `\frac{a}{b}` | $\frac{a}{b}$ |
| Square root | `\sqrt{x}` | $\sqrt{x}$ |
| Greek letters | `\omega`, `\Omega` | $\omega$, $\Omega$ |
| Sum | `\sum_{k=0}^{N-1}x[k]` | $\sum_{k=0}^{N-1}x[k]$ |
| Integral | `\int_a^b x(t)\,dt` | $\int_a^b x(t)\,dt$ |
| Magnitude | `\lvert X\rvert` | $\lvert X\rvert$ |
| Vector | `\mathbf{x}` | $\mathbf{x}$ |
| Upright units | `\mathrm{mV}` | $\mathrm{mV}$ |
| Approximation | `\approx` | $\approx$ |

Braces group the characters belonging to a command. They matter when a subscript or superscript contains more than one character.

## H.3 Align a derivation

The `aligned` environment lines up steps at `&`. A double backslash ends a row.

```text
\begin{aligned}
y[n] &= \frac{x[n]+x[n-1]+x[n-2]}{3} \\
     &= \frac{0.3+0.6+0.9}{3} \\
     &= 0.6\,\mathrm{mV}.
\end{aligned}
```

$$
\begin{aligned}
y[n] &= \frac{x[n]+x[n-1]+x[n-2]}{3} \\
     &= \frac{0.3+0.6+0.9}{3} \\
     &= 0.6\,\mathrm{mV}.
\end{aligned}
$$

The substituted samples have units of millivolts. Each step replaces an expression with an equivalent value. Prose between equations is often easier to follow than one long block.

## H.5 Common mistakes

Pair every delimiter and group multi-character subscripts with braces. Distinguish typesetting from Python: `^` creates a superscript in LaTeX, while Python uses `**` for exponentiation. The website supports math commands through MathJax; document commands such as `\documentclass` belong in a LaTeX document.

See the [symbol guide](i_math_symbols.ipynb) for the meaning behind the commands, and the [MathJax documentation](https://docs.mathjax.org/en/latest/input/tex/index.html) for supported syntax.
